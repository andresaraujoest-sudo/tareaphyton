### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
# Escriba su solución aquí
producto = input("Ingrese el nombre del producto: ")
cantidad = int(input("Ingrese la cantidad: "))
precio = float(input("Ingrese el precio unitario: "))

subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = (subtotal - descuento) + iva

print(f"Producto: {producto}")
print(f"Subtotal: {subtotal:10.2f}")
print(f"Descuento: {descuento:10.2f}")
print(f"IVA: {iva:10.2f}")
print(f"TOTAL: {total:10.2f}")
````

### Ejercicio 5 · Distancia entre dos puntos ★★

Pida las coordenadas de dos puntos (x1, y1) y (x2, y2) y calcule la distancia entre ellos con la fórmula
$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$.
Hágalo de dos formas: con `** 0.5` y con `math.sqrt`. Muestre el resultado con 3 decimales.

````python
# Escriba su solución aquí

import math

# Pidiendo las coordenadas de los dos puntos
x1 = float(input("Ingrese x1: "))
y1 = float(input("Ingrese y1: "))
x2 = float(input("Ingrese x2: "))
y2 = float(input("Ingrese y2: "))

# Forma 1: Usando ** 0.5 (potencia fraccionaria para la raíz)
distancia_1 = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

# Forma 2: Usando math.sqrt
distancia_2 = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

# Mostrando los resultados con 3 decimales
print(f"Distancia (con ** 0.5): {distancia_1:.3f}")
print(f"Distancia (con math.sqrt): {distancia_2:.3f}")
````

### Ejercicio 8 · Reparto de una cuenta ★★

Un grupo de amigos va a cenar. Pida el total de la cuenta, el porcentaje de propina y el número de personas. Muestre:
1. El valor de la propina.
2. El total con propina.
3. Cuánto paga cada persona, redondeado a 2 decimales.
4. Cuánto sobra si cada uno paga la cantidad redondeada **hacia arriba** al dólar entero (use `math.ceil`).

````python
# Escriba su solución aquí

import math

total_cuenta = float(input("Ingrese el total de la cuenta: "))
porcentaje_propina = float(input("Ingrese el porcentaje de propina (%): "))
num_personas = int(input("Ingrese el número de personas: "))

# 1. El valor de la propina
valor_propina = total_cuenta * (porcentaje_propina / 100)

# 2. El total con propina
total_con_propina = total_cuenta + valor_propina

# 3. Cuánto paga cada persona, redondeado a 2 decimales
pago_por_persona = total_con_propina / num_personas
pago_redondeado_2dec = round(pago_por_persona, 2)

# 4. Cuánto sobra si cada uno paga la cantidad redondeada hacia arriba al dólar entero
pago_por_persona_arriba = math.ceil(pago_por_persona)
recaudacion_total = pago_por_persona_arriba * num_personas
dinero_sobrante = recaudacion_total - total_con_propina

print(f"1. Valor de la propina: {valor_propina:.2f}")
print(f"2. Total con propina: {total_con_propina:.2f}")
print(f"3. Cuánto paga cada persona (redondeado): {pago_redondeado_2dec:.2f}")
print(f"4. Cuánto sobra si cada uno paga redondeando hacia arriba (${pago_por_persona_arriba}): {dinero_sobrante:.2f}")
````

### Ejercicio 12 · Ahorro con interés compuesto ★★

Pida el capital inicial, la tasa anual (%) y los años. Calcule el monto final con interés compuesto, $M = C(1 + t)^n$, y la ganancia. Compárelo con el interés simple ($C \cdot t \cdot n$) y muestre la diferencia, todo con `:,.2f`.

````python
# Escriba su solución aquí

capital = float(input("Ingrese el capital inicial (C): "))
tasa_anual = float(input("Ingrese la tasa anual en porcentaje (%): "))
anios = int(input("Ingrese el número de años (n): "))

# Convertir la tasa porcentual a decimal
t = tasa_anual / 100

# Cálculo del monto final con interés compuesto: M = C * (1 + t)**n
monto_compuesto = capital * (1 + t) ** anios
ganancia_compuesta = monto_compuesto - capital

# Cálculo del interés simple: C * t * n (y ganancia total)
interes_simple = capital * t * anios
monto_simple = capital + interes_simple

# Diferencia entre el monto compuesto y el monto simple
diferencia = monto_compuesto - monto_simple

print(f"--- Resultados del Ahorro ---")
print(f"Monto final (Interés compuesto): {monto_compuesto:,.2f}")
print(f"Ganancia (Interés compuesto): {ganancia_compuesta:,.2f}")
print(f"Monto final (Interés simple): {monto_simple:,.2f}")
print(f"Ganancia (Interés simple): {interes_simple:,.2f}")
print(f"Diferencia (Compuesto - Simple): {diferencia:,.2f}")
````

### Ejercicio 16 · Limpiar un registro ★★

Un sistema exportó este registro con espacios y mayúsculas desordenadas:
`registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "`
1. Sepárelo por comas con `split`.
2. Limpie cada parte con `strip`.
3. Arme un diccionario con `cedula`, `nombre` (formato título y un solo espacio entre palabras), `correo` (minúsculas) y `saldo` (`float`).
4. Muéstrelo.

💡 Para dejar un solo espacio: `" ".join(texto.split())`.

````python
registro = "  0104567896 , pérez ANA , ANA.PEREZ@MAIL.COM , 150.50  "

# 1. Separarlo por comas con split()
partes = registro.split(",")

# 2. Limpiar cada parte con strip()
cedula = partes[0].strip()
nombre_sucio = partes[1].strip()
correo_sucio = partes[2].strip()
saldo_sucio = partes[3].strip()

# Formatear el nombre (título y un solo espacio entre palabras)
nombre = " ".join(nombre_sucio.split()).title()

# Formatear el correo (minúsculas)
correo = correo_sucio.lower()

# Convertir el saldo a float
saldo = float(saldo_sucio)

# 3. Armar un diccionario con los datos limpios
datos_usuario = {
    "cedula": cedula,
    "nombre": nombre,
    "correo": correo,
    "saldo": saldo
}

# 4. Muéstrelo
print(datos_usuario)
````

### Ejercicio 17 · Usuario y correo institucional ★★

Con `nombre = "María José Torres Ríos"` genere:
1. El **usuario**: la inicial del primer nombre + el primer apellido, en minúsculas y sin tildes → `mtorres`.
2. El **correo**: `usuario@tecazuay.edu.ec`.
3. Las **iniciales** en mayúsculas → `MJTR`.

💡 Use `split()`, índices, `lower()` y `replace()` para quitar las tildes.

````python
nombre = "María José Torres Ríos"

# Separar las palabras del nombre completo
palabras = nombre.split()

# 1. El usuario: inicial del primer nombre + primer apellido (minúsculas y sin tildes)
primera_letra = palabras[0][0].lower() \
    .replace("á", "a").replace("é", "e").replace("í", "i").replace("ó", "o").replace("ú", "u")

primer_apellido = palabras[2].lower() \
    .replace("á", "a").replace("é", "e").replace("í", "i").replace("ó", "o").replace("ú", "u")

usuario = primera_letra + primer_apellido

# 2. El correo institucional
correo = f"{usuario}@tecazuay.edu.ec"

# 3. Las iniciales en mayúsculas (primera letra de cada palabra)
iniciales = (palabras[0][0] + palabras[1][0] + palabras[2][0] + palabras[3][0]).upper() \
    .replace("Á", "A").replace("É", "E").replace("Í", "I").replace("Ó", "O").replace("Ú", "U")

print(f"1. Usuario: {usuario}")
print(f"2. Correo: {correo}")
print(f"3. Iniciales: {iniciales}")
````

### Ejercicio 18 · Ocultar un número de cuenta ★★

Con `cuenta = "2203456789012345"`:
1. Muestre la cuenta oculta, dejando visibles solo los 4 últimos dígitos: `************2345`.
2. Muéstrela en grupos de 4 separados por espacios: `2203 4567 8901 2345` (use slicing).
3. Muestre cuántas veces aparece el dígito `"0"` y en qué posición aparece el primer `"9"` (use `count` y `find`).

````python
cuenta = "2203456789012345"

# 1. Muestre la cuenta oculta, dejando visibles solo los 4 últimos dígitos
cuenta_oculta = "*" * (len(cuenta) - 4) + cuenta[-4:]

# 2. Muéstrela en grupos de 4 separados por espacios usando slicing
g1 = cuenta[0:4]
g2 = cuenta[4:8]
g3 = cuenta[8:12]
g4 = cuenta[12:16]
cuenta_espacios = f"{g1} {g2} {g3} {g4}"

# 3. Cuantas veces aparece el dígito "0" y en qué posición aparece el primer "9"
cantidad_ceros = cuenta.count("0")
posicion_nueve = cuenta.find("9")

print(f"1. Cuenta oculta: {cuenta_oculta}")
print(f"2. Cuenta en grupos de 4: {cuenta_espacios}")
print(f"3. Veces que aparece el '0': {cantidad_ceros}")
print(f"   Posición del primer '9': {posicion_nueve}")
````

### Ejercicio 22 · Inventario con diccionario ★★

Con `stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}`:
1. Se vendieron 5 mouses: actualice el stock.
2. Llegó un producto nuevo: agregue `"webcam": 15` con `update`.
3. Se dejó de vender el monitor: elimínelo con `pop` y muestre cuántas unidades tenía.
4. Consulte `"impresora"` con `get` para que muestre `0` si no existe.
5. Muestre el total de unidades, los productos en orden alfabético y el producto con más unidades (`max(stock, key=stock.get)`).

````python
stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}

# 1. Se vendieron 5 mouses: actualice el stock
stock["mouse"] -= 5

# 2. Llegó un producto nuevo: agregue "webcam": 15 con update
stock.update({"webcam": 15})

# 3. Se dejó de vender el monitor: elimínelo con pop y muestre cuántas unidades tenía
unidades_monitor = stock.pop("monitor")

# 4. Consulte "impresora" con get para que muestre 0 si no existe
stock_impresora = stock.get("impresora", 0)

# 5. Muestre el total de unidades, los productos en orden alfabético y el producto con más unidades
total_unidades = sum(stock.values())
productos_ordenados = sorted(stock.keys())
producto_mas_unidades = max(stock, key=stock.get)

print(f"1. Stock actualizado (mouse): {stock['mouse']}")
print(f"2. Inventario tras update: {stock}")
print(f"3. Unidades del monitor eliminadas: {unidades_monitor}")
print(f"4. Consulta 'impresora' con get: {stock_impresora}")
print(f"5. Total de unidades: {total_unidades}")
print(f"   Productos en orden alfabético: {productos_ordenados}")
print(f"   Producto con más unidades: {producto_mas_unidades}")
````

### Ejercicio 24 · Operaciones con conjuntos ★★

Estudiantes inscritos en tres talleres:
```python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t = {"Luis", "Carla", "Pedro", "Diego"}
ia_t = {"Ana", "Pedro", "Diego"}

# 1. Quiénes están en Python y en Excel (intersección)
python_y_excel = sorted(python_t & excel_t)

# 2. Quiénes están en los tres talleres (intersección múltiple)
en_los_tres = sorted(python_t & excel_t & ia_t)

# 3. Quiénes están en Python pero no en IA (diferencia)
python_no_ia = sorted(python_t - ia_t)

# 4. Quiénes están en Python o en Excel, pero no en ambos (diferencia simétrica ^)
python_o_excel_exclusivo = sorted(python_t ^ excel_t)

# 5. Si todos los de IA están también en Python (issubset)
ia_en_python = ia_t.issubset(python_t)

# 6. El total de estudiantes distintos (unión total)
estudiantes_distintos = len(python_t | excel_t | ia_t)

print(f"1. En Python y Excel: {python_y_excel}")
print(f"2. En los tres talleres: {en_los_tres}")
print(f"3. En Python pero no en IA: {python_no_ia}")
print(f"4. En Python o Excel (exclusivo): {python_o_excel_exclusivo}")
print(f"5. ¿Todos los de IA están en Python?: {ia_en_python}")
print(f"6. Total de estudiantes distintos: {estudiantes_distintos}")
````

4. Explique en un comentario por qué en un sistema bancario se prefiere `Decimal`.

````python
from decimal import Decimal

# 1. Muestre el resultado de 0.1 + 0.2 y compruebe si es igual a 0.3. ¿Qué ocurre?
suma_flotantes = 0.1 + 0.2
es_igual = (suma_flotantes == 0.3)
print(f"1. 0.1 + 0.2 con floats da: {suma_flotantes}")
print(f"   ¿Es igual a 0.3?: {es_igual}")
# Explicación: Ocurre un error de precisión de punto flotante debido a cómo la 
# computadora almacena los números en binario (IEEE 754), resultando en 0.30000000000000004.

# 2. Repita la operación con Decimal("0.1") + Decimal("0.2")
suma_decimales = Decimal("0.1") + Decimal("0.2")
es_igual_decimal = (suma_decimales == Decimal("0.3"))
print(f"2. Con Decimal: {suma_decimales}")
print(f"   ¿Es igual a Decimal('0.3')?: {es_igual_decimal}")

# 3. Calcule el interés de $1 234.56 al 3.75 % con Decimal y redondéelo a centavos con quantize
capital = Decimal("1234.56")
tasa = Decimal("0.0375")
interes_exacto = capital * tasa
interes_redondeado = interes_exacto.quantize(Decimal("0.01"))
print(f"3. Interés calculado y redondeado a centavos: ${interes_redondeado}")

# 4. Explique en un comentario por qué en un sistema bancario se prefiere Decimal
# ---------------------------------------------------------------------------------
# COMENTARIO EXPLICATIVO:
# En los sistemas bancarios y financieros se prefiere utilizar la clase 'Decimal' en 
# lugar de floats tradicionales porque los floats introducen pequeños errores de redondeo 
# por las limitaciones de representación binaria. En transacciones financieras, acumular 
# esos micro-errores puede generar descuadres graves de dinero en cuentas o auditorías, 
# mientras que 'Decimal' opera con aritmética exacta basada en base 10 (como las matemáticas humanas).
# ---------------------------------------------------------------------------------
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"

# 1. Separe la línea por "|"
partes = linea.split("|")
nombre_raw = partes[0]
cedula = partes[1]
cuenta1_str = partes[2]
cuenta2_str = partes[3]

# 2. Cree un diccionario con nombre (formato título), cedula y cuentas
nombre_titulo = nombre_raw.title()

tipo1, saldo1 = cuenta1_str.split("=")
tipo2, saldo2 = cuenta2_str.split("=")

cuentas_dict = {
    tipo1: float(saldo1),
    tipo2: float(saldo2)
}

socio = {
    "nombre": nombre_titulo,
    "cedula": cedula,
    "cuentas": cuentas_dict
}

# 3. Muestre la ficha con el formato solicitado
tipos_cuenta = f"{tipo1}, {tipo2}"
saldo_total = sum(socio["cuentas"].values())

print(f"Socio: {socio['nombre']} ({socio['cedula']})")
print(f"Tipos de cuenta: {tipos_cuenta}")
print(f"Saldo total: ${saldo_total:,.2f}")
````
